# Build a TRAIN MANIFEST

```python
import os
import glob
import pandas as pd

TRAIN_DATA = "/kaggle/input/datasets/samasiayushman/small-model-track/Training/Training/data"

modalities = [
    "Depth_Color",
    "IR",
    "Thermal",
    "IMU",
    "Radar",
    "Skeleton"
]

rows = []

for action in sorted(os.listdir(os.path.join(TRAIN_DATA, "Skeleton"))):

    action_path = os.path.join(TRAIN_DATA, "Skeleton", action)

    if not os.path.isdir(action_path):
        continue

    label = int(action.split("_")[0])

    for user in sorted(os.listdir(action_path)):

        user_path = os.path.join(action_path, user)

        if not os.path.isdir(user_path):
            continue

        for trial in sorted(os.listdir(user_path)):

            trial_path = os.path.join(user_path, trial)

            if not os.path.isdir(trial_path):
                continue

            row = {
                "action": action,
                "label": label,
                "user": user,
                "trial": trial
            }

            # Skeleton
            row["skeleton"] = len(
                glob.glob(
                    os.path.join(
                        trial_path,
                        "predictions",
                        "*.json"
                    )
                )
            )

            # Other modalities
            for modality in modalities:

                if modality == "Skeleton":
                    continue

                p = os.path.join(
                    TRAIN_DATA,
                    modality,
                    action,
                    user,
                    trial
                )

                if modality == "IMU":
                    count = len(glob.glob(os.path.join(p, "*.csv")))

                elif modality == "Radar":
                    count = len(glob.glob(os.path.join(p, "*.csv")))

                else:
                    count = len(glob.glob(os.path.join(p, "*")))

                row[modality.lower()] = count

            rows.append(row)

manifest = pd.DataFrame(rows)

print("Number of training trials:", len(manifest))
print()
print(manifest.head())
```

```python
import json
import numpy as np
from pathlib import Path


def load_skeleton(trial_path):
    """
    Load all skeleton JSON frames from one trial.

    Returns:
        keypoints:      [T, 17, 3]
        scores:         [T, 17]
    """

    trial_path = Path(trial_path)
    prediction_dir = trial_path / "predictions"

    json_files = sorted(prediction_dir.glob("*.json"))

    keypoints = []
    scores = []

    for json_file in json_files:

        with open(json_file, "r") as f:
            data = json.load(f)

        # No detected person
        if len(data) == 0:
            keypoints.append(np.zeros((17, 3), dtype=np.float32))
            scores.append(np.zeros(17, dtype=np.float32))
            continue

        # First detected person
        person = data[0]

        kp = np.asarray(
            person["keypoints"],
            dtype=np.float32
        )

        score = np.asarray(
            person["keypoint_scores"],
            dtype=np.float32
        )

        keypoints.append(kp)
        scores.append(score)

    if len(keypoints) == 0:
        return (
            np.empty((0, 17, 3), dtype=np.float32),
            np.empty((0, 17), dtype=np.float32)
        )

    keypoints = np.stack(keypoints)
    scores = np.stack(scores)

    return keypoints, scores
```

```python
import json
import numpy as np


def load_skeleton(trial_path):

    pred_dir = trial_path / "predictions"

    json_files = sorted(pred_dir.glob("*.json"))

    frames = []
    scores = []

    for json_file in json_files:

        with open(json_file, "r") as f:
            data = json.load(f)

        # Empty frame
        if len(data) == 0:
            keypoints = np.zeros((17, 3), dtype=np.float32)
            kp_scores = np.zeros(17, dtype=np.float32)

        else:
            # First / only person
            person = data[0]

            keypoints = np.asarray(
                person["keypoints"],
                dtype=np.float32
            )

            kp_scores = np.asarray(
                person["keypoint_scores"],
                dtype=np.float32
            )

        frames.append(keypoints)
        scores.append(kp_scores)

    skeleton = np.stack(frames)
    scores = np.stack(scores)

    return skeleton, scores
```
